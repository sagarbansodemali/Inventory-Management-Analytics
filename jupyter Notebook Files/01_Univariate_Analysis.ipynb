```python

# This function returns the categorical univariate value_counts & pie chart of column.

def cat_Univariate_pie (column_name) :
    Unique_Counts = column_name.value_counts()
    Pie_Chart_of_column = column_name.value_counts().plot(kind="pie",autopct="%0.2f")
    print (Unique_Counts)
    return  Pie_Chart_of_column
    
```

```python
cat_Univariate_pie(inventory_movement["Movement_Type"])
```

```python
inventory_movement['Movement_Quantity'].describe()
```

```python
sns.distplot (x=inventory_movement['Movement_Quantity'])
```

```python
cat_Univariate_pie(inventory_movement["Destination"])
```
